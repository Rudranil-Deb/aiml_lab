Problem 2: The Water Jug Problem (Solved using BFS)
Problem Statement:
You are given two jugs: Jug A with a capacity of 4 litres, and Jug B with a capacity of 3 litres. Neither jug has any markings on it. Using only the following operations — (i) fill a jug completely, (ii) empty a jug completely, (iii) pour water from one jug into the other until either the first jug is empty or the second jug is full — determine the minimum sequence of steps needed to measure out exactly 2 litres of water in any one jug.

```python
from collections import deque

# Capacities of the jugs
capacity_A = 4
capacity_B = 3


# Generate all possible next states
def get_next_states(state):
    A, B = state
    states = []

    # 1. Fill Jug A
    states.append(((4, B), "Fill Jug A"))

    # 2. Fill Jug B
    states.append(((A, 3), "Fill Jug B"))

    # 3. Empty Jug A
    states.append(((0, B), "Empty Jug A"))

    # 4. Empty Jug B
    states.append(((A, 0), "Empty Jug B"))

    # 5. Pour Jug A into Jug B
    amount = min(A, 3 - B)
    states.append(((A - amount, B + amount), "Pour A -> B"))

    # 6. Pour Jug B into Jug A
    amount = min(B, 4 - A)
    states.append(((A + amount, B - amount), "Pour B -> A"))

    return states


# BFS function
def bfs():
    start = (0, 0)

    queue = deque([start])
    visited = {start}

    # Store the previous state and operation
    parent = {start: None}
    action = {start: None}

    while queue:
        current = queue.popleft()

        A, B = current

        # Goal: exactly 2 litres in either jug
        if A == 2 or B == 2:

            path = []

            while current is not None:
                path.append((current, action[current]))
                current = parent[current]

            path.reverse()
            return path

        # Generate next states
        for next_state, operation in get_next_states(current):

            if next_state not in visited:
                visited.add(next_state)
                parent[next_state] = current
                action[next_state] = operation
                queue.append(next_state)

    return None


# Solve the problem
solution = bfs()

# Print the solution
print("Minimum sequence of steps:\n")

for i, (state, operation) in enumerate(solution):

    if i == 0:
        print(f"Step {i}: {state} - Initial State")
    else:
        print(f"Step {i}: {state} - {operation}")

print("\nMinimum number of steps:", len(solution) - 1)
```
